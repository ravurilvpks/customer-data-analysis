```python
df["age_cat"]=pd.cut(
    df["age"],
    bins=[0,18,30,45,60,100],
    labels=["0-18","19-30","31-45","46-60","61+"])
```

```python
df[["age","age_cat"]].head()
```

```python
df.groupby("age_cat",observed=True)["quantity"].sum()
```
